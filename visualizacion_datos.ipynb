```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = pd.read_csv('top3_mis_artistas.csv')
```

#Configuracion de estilo Sea Born

```python
sns.set_theme(style="whitegrid")
```

#Grafico de distribucion de popularidad

```python
plt.figure(figsize=(10,6))
sns.histplot(data['Duracion_seg'], kde=True, bins=30)
plt.title('Distribucion de duracion de las canciones')
plt.xlabel('Duracion [seg]')
plt.ylabel('Frecuencia')
plt.show()
```
